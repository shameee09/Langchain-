```python

# 1. Imports
from getpass import getpass
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser

# 2. Gemini API Key
GOOGLE_API_KEY = getpass("Enter Gemini API Key: ")

# 3. Setup Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0,
    google_api_key=GOOGLE_API_KEY
)

# 4. Create Prompt
prompt = PromptTemplate.from_template(
    "List 5 programming languages, comma-separated."
)

# 5. Output Parser
parser = CommaSeparatedListOutputParser()

# 6. Create LCEL Chain
chain = prompt | llm | parser

# 7. Execute
response = chain.invoke({})

print("Programming Languages:")
print(response)
```

```python

# 1. Imports
from getpass import getpass
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field

# 2. Gemini API Key
GOOGLE_API_KEY = getpass("Enter Gemini API Key: ")

# 3. Setup Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0,
    google_api_key=GOOGLE_API_KEY
)

# 4. Define Pydantic Model
class ProductInfo(BaseModel):
    name: str = Field(description="Name of the product")
    price: float = Field(description="Price in INR")

# 5. Pydantic Output Parser
parser = PydanticOutputParser(
    pydantic_object=ProductInfo
)

# 6. Create Prompt
prompt = PromptTemplate(
    template="Extract product name and price from: {text}\n{format_instructions}",
    input_variables=["text"],
    partial_variables={
        "format_instructions": parser.get_format_instructions()
    },
)

# 7. Create LCEL Chain
chain = prompt | llm | parser

response = chain.invoke({
    "text": "The Redmi Note 12 is available for ₹14,999."
})

print(response)
print("Product Name:", response.name)
print("Price:", response.price)
```

```python

# 1. Imports
from getpass import getpass
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_classic.output_parsers import (
    ResponseSchema,
    StructuredOutputParser
)

# 2. Gemini API Key
GOOGLE_API_KEY = getpass("Enter Gemini API Key: ")

# 3. Setup Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0,
    google_api_key=GOOGLE_API_KEY
)

# 4. Define Response Schemas
schemas = [
    ResponseSchema(
        name="company",
        description="Name of the company"
    ),
    ResponseSchema(
        name="founder",
        description="Name of the founder"
    ),
]

# 5. Structured Output Parser
parser = StructuredOutputParser.from_response_schemas(schemas)

# 6. Create Prompt
prompt = PromptTemplate(
    template="Extract company and founder from the text: {text}\n{format_instructions}",
    input_variables=["text"],
    partial_variables={
        "format_instructions": parser.get_format_instructions()
    }
)

# 7. Create LCEL Chain
chain = prompt | llm | parser

# 8. Execute
response = chain.invoke({
    "text": "Hope AI was founded by Ramisha Rani in Tamil Nadu."
})

print(response)
```
